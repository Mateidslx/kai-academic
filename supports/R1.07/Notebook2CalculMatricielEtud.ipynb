# **Notebook 2 : Calcul matriciel en Python**
Pour le calcul matriciel en Python, on utilise habituellement la bibliothèque **Numpy** (pour Numerical Python) dont la structure de données fondamentale est le tableau multidimensionnel (1D, 2D, 3D...), communément appelé **array**, **ndarray** ou même **tableau numpy**.  
En deux dimensions, un **ndarray** est parfaitement adapté à la représentation des matrices, NumPy proposant par ailleurs les **opérations matricielles** classiques : addition, multiplication par un scalaire, produit d’Hadamard, produit matriciel, transposée, inverse, etc.

Dans ce notebook, nous adopterons une représentation plus simple, à l’aide de **listes de listes**. Cette approche est moins performante que celle offerte par NumPy, les calculs étant plus lents, mais elle présente l’intérêt de nous permettre de programmer nous-mêmes les opérations matricielles, pour mieux en comprendre les mécanismes et la logique.

### **Deux remarques sur les tableaux Numpy et les listes de listes**
1. Il est aisé de passer d'une liste de listes à un tableau Numpy, on utilise pour cela la fonction `np.array()`  
   Dans l'exemple ci-dessous, la matrice $A= \left( \begin{matrix} 1&2&-1\\0&3&1\end{matrix} \right)$  est d'abord représentée par la **liste de listes A** (les éléments de la liste sont les **lignes** de la matrice), qui est ensuite transformée en **tableau Numpy tabA**.

```python
# Création de la liste de liste
A = [[1, 2, -1], [0, 3, 1]]
print('Liste de listes A : \n', A)

# Transformation en tableau numpy ou ndarray
import numpy as np
tabA = np.array(A)
print('Tableau numpy tabA : \n', tabA)
```

2. Les accès aux valeurs de la matrice, sont similaires : `A[i][j]` pour la liste de liste, `tabA[i,j]` pour le tableau numpy

```python
# accès à l'élément de la 2ème ligne et première colonne
print(A[1][0])
print(tabA[1,0])
```

## **Remarque préliminaire : initialisation d'une liste**
Sous Python, il n'est pas nécessaire de déclarer explicitement les variables que l'on manipule. La déclaration et l'initialisation se font en même temps.

Par exemple, lorsque nous écrivons `x = 3`, Python détermine automatiquement que `x` est de type entier, alloue l'espace mémoire nécessaire et assigne la valeur 3 à la variable `x`. Cela reflète l'une des caractéristiques des langages de haut niveau à typage dynamique.

Cependant, Python ne peut pas tout "deviner". Par exemple, si on manipule une liste de n éléments, on ne pourra pas affecter directement une valeur à un élément d'indice i si cette liste n’a pas été préalablement remplie avec suffisamment d’éléments.

* L'instruction `liste = []` crée bien une liste vide, sans générer d'erreur.
* Mais l'instruction `liste[2] = 3` renverra une erreur si la liste n'existe pas encore ou ne contient pas déjà au moins trois éléments (c’est-à-dire d'indices 0, 1 et 2).

Une solution consiste à initialiser la liste avec une taille adéquate, ce qui permet ensuite de modifier les éléments par indice, comme dans un **tableau de taille fixe**.
`

Les exemples suivants montrent des initialisations de listes où les éléments sont initialement définis à `None`, ce qui représente une absence de valeur. Cette approche permet de réserver l’espace et d’indiquer que ces éléments  pourront être remplis ultérieurement dans le programme.

```python
liste = [None]*10
liste
```

```python
liste = [None for i in range(10)]
liste 
```

```python
liste =[]
for i in range(10):
    liste.append(None)
liste 
```

# **Produit matriciel**
# Préliminaire : copie superficielle et profonde d'une liste

```python
# Illustration
liste1 = [6,7,8]
liste2 = liste1
print(liste2)
```

```python
# Modification de liste2
liste2[1] = 0
print(liste2)
```

```python
# Mais liste1 a également été modifié
print(liste1)
```

##  Explication ?
Une liste en Python contient l'adresse mémoire où sont stockés ses éléments. Vous verrez cette notion en semestre 2, avec le concept de pointeur.

En copiant **liste1** dans **liste2**, ce n'est pas une copie des éléments eux-mêmes qui se fait, mais de **l'adresse mémoire** où ces éléments sont stockés. Ainsi, les deux listes référencent les mêmes objets en mémoire.

La fonction intégrée **id()** retourne l'adresse mémoire de l'objet. Dans l'exemple ci-dessous, nous pouvons vérifier que les deux objets sont identiques, car ils partagent la même adresse mémoire.

```python
print (id(liste1), id(liste2))
```

**liste1** et **liste2** font référence au même emplacement mémoire, ce qui signifie que la copie de **liste1** dans **liste2** est **superficielle**. Une modification d’un élément de **liste2** produit la même modification dans **liste1**, puisque les éléments de **liste1** et **liste2** sont stockés aux mêmes endroits en mémoire.

* **Particularité de la méthode **copy()** sur les listes, permettant de faire une copie "superficielle"**

La méthode **copy()** crée une nouvelle liste, qui n'a pas le même identifiant, mais qui contient des références vers les **mêmes** objets que l'original.

Avec **copy()**, l'affectation va se faire au niveau de **chaque élément** de la liste. 
* Si cet élément est un "objet", c'est l'adresse qui est copiée (comme dans l'exemple précédent) : les deux éléments sont toujours liés par leur **adresse**.
* Si cet élément est un entier, c'est la valeur qui est copiée et les deux éléments sont **dissociés** (un entier est toujours copié par valeur, et pour un entier, la fonction id() n'a plus de sens).

Ci-dessous deux exemples :
#### **Exemple1 : avec des listes de listes**

```python
lisliste1 = [[2,3],[4,5]]
lisliste2 = lisliste1.copy()
```

```python
# On vérifie que les deux objets ont des identifiants différents
print("identifiant de lisliste1 : ", id(lisliste1))
print("identifiant de lisliste2 : ", id(lisliste2))
```

Les deux listes n'ont pas les mêmes **identifiants**, en revanche leurs éléments qui sont des objets (listes) font référence aux mêmes **adresses**. 

**Par exemple pour le premier élément :**

```python
print("identifiant du premier élément de lisliste1 : ", id(lisliste1[0]))
print("identifiant du premier élément de lisliste2 : ", id(lisliste2[0]))
```

```python
# Modification du premier élément de lisliste2
lisliste1[1][0] = 11
print("lisliste1 : ", lisliste1)
print("lisliste2 : ", lisliste2)
```

#### **Exemple2 : avec des listes d'entiers**

```python
liste1 = [6,7,8]
liste10 = liste1.copy()
print(liste10)
```

```python
liste10[1] = 77
print("liste1 : ", liste1)
print("liste10 : ", liste10)
```

Ce n'est plus l'adresse, mais la valeur qui est copiée et les deux éléments sont dissociés

## 2.3 Fonction colonne
Créer une fonction **colonne(M, j)** prenant en entrée une matrice **M** (liste de liste) et **j** un indice de colonne et renvoyant **la jème colonne de la matrice M** 

```python
def colonne(M,j) : 
    liste = []
    for i in range(len(M)) : 
        liste.append(M[i][j])
    return liste
```

## 2.4 Fonction produit ligne-colonne
Créer une fonction **prod_ligne_col(ligne,colonne)** :
* prenant en entrée une liste **ligne** (représentant un vecteur ligne) et une liste **colonne** (représentant un vecteur colonne, **ligne** et **colonne** ayant le même nombre d'éléments.
* renvoyant le **produit "ligne-colonne"** des deux vecteurs.

```python
def prod_ligne_col(ligne,colonne) : 
    prod = 0
    for k in range(len(ligne)) : 
        prod += ligne[k] * colonne[k]
    return prod
```

**Appel de la fonction prod_ligne_colo**

```python
l = [2, 5, -1]
c = [-1, 3, 4]
print("Produit :", prod_ligne_col(l,c))
```

# 3. Produit matriciel 
Ecrire une fonction **prod_mat(mat1,mat2)** :
* prenant en entrée deux listes de listes **mat1** et **mat2** représentant des listes de listes
* renvoyant le **produit** de ces deux matrices sous forme de liste de liste

*On supposera dans un premier temps que les matrices ont les bonnes dimensions et que le produit est possible*

```python
def prod_mat(mat1,mat2) : 
    prod = [[0 for j in range(len(mat2[0]))] for i in range(len(mat1))]
    for i in range(len(mat1)):
        for j in range(len(mat2[0])):
            prod[i][j] = prod_ligne_col(ligne(mat1,i), colonne(mat2,j))
    return prod
```

**Appel de la fonction prod_mat**

```python
print("M1xM2 :", prod_mat(M1,M2))
```
