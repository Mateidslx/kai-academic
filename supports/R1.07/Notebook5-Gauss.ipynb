# **Pivot de Gauss**

## **Transformation d'un système en système triangulaire**
On considère le système suivant $\begin{cases}
2x + 5y + 3z = 104 \\
x + 3y + 2z = 64 \\
3x + 2y + 2z = 55
\end{cases}$
1. Sur votre brouillon, en utilisant la méthode du pivot de Gauss, transformer le système en système triangulaire.
2. Le système ci-dessus peut être représenté matriciellement : $\begin{pmatrix}
2 & 5 & 3 \\
1 & 3 & 2 \\
3 & 2 & 2
\end{pmatrix}
\begin{pmatrix}
x \\
y \\
z
\end{pmatrix}
=
\begin{pmatrix}
104 \\
64 \\
55
\end{pmatrix}$  
En Python :
    * la matrice 
$\begin{pmatrix}
2 & 5 & 3 \\
1 & 3 & 2 \\
3 & 2 & 2
\end{pmatrix}$ peut être représentée par une liste de liste **m = [[2, 5, 3], [1, 3, 2], [3, 2, 2]]**
    * le vecteur colonne $\begin{pmatrix}
104 \\
64 \\
55
\end{pmatrix}$ par la liste **v = [104, 64, 55]**.
  
Écrire une fonction **pivotgauss(m, v)** prenant en paramètres une liste de liste **m** et une liste **v** représentant un système et renvoyant la liste de liste et la liste représantant le système triangulaire équivalent.  
On utilisera, bien sûr l'algorithme du Pivot de Gauss.  
   **Par exemple :** $\begin{pmatrix}
2 & 5 & 3 \\
1 & 3 & 2 \\
3 & 2 & 2
\end{pmatrix}
\begin{pmatrix}
x \\
y \\
z
\end{pmatrix}
=
\begin{pmatrix}
104 \\
64 \\
55
\end{pmatrix}
\quad \Leftrightarrow \quad
\begin{pmatrix}
1 & 3 & 2 \\
0 & -1 & -1 \\
0 & 0 & 3
\end{pmatrix}
\begin{pmatrix}
x \\
y \\
z
\end{pmatrix}
=
\begin{pmatrix}
64 \\
-24 \\
31
\end{pmatrix}$
   
La fonction **pivotgauss(m, v)** doit renvoyer la liste de liste **t = [[1,3,2],[0,-1,-1],[0,0,3]]** et la liste **v = [64,-24,31]**.

## **Algorithme de remontée**

On considère le système triangulaire suivant : 
$\left\{
\begin{array}{r@{\,}r@{\,}r@{\,}r@{\,}l}
x & + y & + z & - t & = 2 \\
   & + 2y & + z & + 2t & = -3 \\
   &      & + 3z & - t & = 4 \\
   &      &      & - 2t & = 2
\end{array}
\right.$
  
1. Résoudre ce système sur votre brouillon
2. Écrire un programme permettant de définir la fonction **remontee(t,v)**, prenant en paramètres une liste de liste **t** et et une liste **v** représentant le système et renvoyant une liste **x** représentant un vecteur colonne $X$, solution de l’équation : $TX = V$  
**Par exemple**, pour le système ci-dessus :
    * **t = [[1,3,2],[0,-1,-1],[0,0,3]]** représente $T = \begin{pmatrix}
1 & 3 & 2 \\
0 & -1 & -1 \\
0 & 0 & 3
\end{pmatrix}$
    * **v = [64,-24,31]** représente $V = \begin{pmatrix}
64 \\
-24 \\
31
\end{pmatrix}$

    **remontee(t,v)**  devra renvoyer **x = [2.33,13.67,10.33]** représentant
    $X = \begin{pmatrix}
2.33 \\
13.67 \\
10.33
\end{pmatrix}$  
     solution de l’équation : $𝑇𝑋=𝑉$
