## Éléments de différentiation des conteneurs
* **Ordonné** (ordered) ?
* **Indexable** ? 
* **Immuable** (immutable) ? 

### **Ordonné ?**
#### **Les listes et les tuples sont ordonnés**

```python
# Les listes sont ordonnées
a1 = [2, 'az', 5]
a2 = [2, 5, 'az']
a1 == a2
```

```python
# Les tuples sont ordonnées
b1 = (2, 'az', 5)
b2 = (2, 5, 'az')
b1 == b2
```

#### **Ni les dictionnaires, ni les ensembles ne sont ordonnés**
Ce qui justifie l'utilisation d'accolades **{ }** qui, nous l'avons vu en mathématiques, sont utilisées pour des **ensembles**, c'est-à-dire des collections non ordonnées.

```python
# Les dictionnaires ne sont pas ordonnés 
c1 = {'entier1':2,'texte':'az','entier2':3}
c2 = {'entier1':2,'entier2':3,'texte':'az'}
c1 == c2
```

```python
# Les ensembles ne sont pas ordonnés
d1 = {2,'az',5}
d2 = {2,5,'az'}
d1 == d2
```

### **Indexable ?**
En **mathématique**, quand un objet *discret** est ordonné, il est possible de **numéroter** ses éléments, on dit que cet objet est **dénombrable**.

En **informatique**, on dira plutôt que l'objet est **indexable** et en général, un objet **ordonné** est **indexable**.

En mathématique, la première valeur de numérotation est **1**. En informatique, cela dépend des langages, en Python, le premier élément a pour indice **0**. 

*\* un objet non "discret" serait par exemple un intervalle, qui n'est pas dénombrable.*

#### **Les listes et les tuples sont indexables** (et ordonnés)

```python
# Affichage du premier élément de la liste a = [2,'az',5]
print(a[0])
```

```python
# Affichage du premier élément du tuple b = (2,'az',5)
print(b[0])
```

#### **Les set et les dictionnaires ne sont pas indexables** (ni ordonnés)

```python
# Affichage du premier élément du dictionnaire
print(c[0]) 
```

```python
# Affichage du premier élément de l’ensemble d = {2,'az',5}
print(d[0])
```

#### **Remarque 1 : indexable vs subscriptable**
Le message d'erreur n'est pas le même dans chacune des situations précédentes. À la différence du **set**, le **dictionnaire**, bien que non indexable, est **subscriptable**, ce qui signifie qu'il supporte une opération du genre **x[k]**.   
Dans notre cas, avec le dictionnaire **c = {'entier1' : 2, 'texte' : 'az', 'entier2' : 3}**, l'instruction `c['entier1']` ne renverrait pas d'erreur, **c est donc subscriptable**.   
Et s'il y avait eu une clé égale à 0 dans le dictionnaire **c**, nous n'aurions pas eu d'erreur, mais dans ce cas-là, `c[0]` aurait renvoyé la valeur de **clé 0** et non la première valeur de c (c n'étant pas ordonné, il n'a pas de **première** valeur).  
**Deux illustrations ci-dessous**

```python
c['entier1']
```

```python
dic = {'entier1':2, 0:'az', 'entier2':3}
print(dic[0])
```

```python
print(dic[1])
```

#### **Remarque 2 : accès au dernier élément**
La numérotation démarrant à 0, le dernier élément d'un conteneur indexable `cont` a donc pour indice `len(cont)-1`.  
En Python, on peut se contenter de préciser `-1`, au lieu de `len(cont)-1`. 

```python
# Les listes sont ordonnées
a = [2, 'az', 5]
print(a[len(a)-1])
print(a[-1])
```

### **Muables ?**
Les tuples, à la différence des listes, des sets et des dictionnaires, ne sont pas muables (ou mu**T**ables)

Dans un objet **muable** :
* on peut ajouter ou enlever des éléments
* les éléments peuvent être réaffectés

#### **Les tuples sont immuables** 

```python
# Les tuples sont immuables 
b=(2,'az',5)
b[2]=4
```

**Commentaire sur l'erreur** : il n'est pas possible de modifier une valeur d'un tuple (ici, la troisième)
#### **Les listes, les sets et les dictionnaires sont muables**

```python
# Modification d’une liste
a=[2,'az',5]
a[2] = 4
print(a)
```

```python
# Modification d’un dictionnaire
c={'entier1':2,'texte':'az','entier2':5}
c['entier2'] = 4
print(c)
```

La modification est un peu plus compliquée pour le **set**

```python
# Modification d’un ensemble
d={2,'az',5}
d.add(4)
print(d)
d.remove(5)
print(d)
```

# Complément 1 : **Parcours de conteneurs indexables (listes, tuples)**
Au lieu de parcourir directement les **valeurs** du conteneur indexable (**for val in conteneur :**), il est possible de parcourir ses **indices**. On utilise pour cela un objet de type **range**, qui permet de générer des suites arithmétiques.
## **Objets range**
Dans les exemples ci-dessous, pour visualiser les suites générées par les **range**, nous les transformons en listes.

```python
# S’il n’y a qu’un seul paramètre, celui-ci correspond au nombre d’entiers générés à partir de 0. 
# La dernière valeur n'est donc pas atteinte
list(range(3))
```

```python
# S’il y a deux paramètres, ceux-ci correspondent aux valeurs de départ et d’arrivée
# la dernière valeur n’est pas atteinte
list(range(4,10))
```

```python
# S’il y a trois paramètres, le troisième correspond au « pas » (raison de la suite arithmétique) 
list(range(4,10,2))
```

```python
# S’il y a trois paramètres, le troisième correspond au « pas » (raison de la suite arithmétique) 
list(range(10,4,-2))
```

## **Parcours de conteneurs indexables**
Les deux boucles suivantes sont donc équivalentes

```python
a = [2, 'az', 5]
for i in range(len(a)):
    print(a[i])
```

```python
a = [2,'az',5]
for val in a:
    print(val)
```

# Complément 2 : **Parcours de dictionnaires**

Nous avons vu qu'un dictionnaire est itérable, mais non indexable.   
On ne peut donc pas parcourir un dictionnaire par des indices qui prendraient des valeurs dans un range.   

```python
c = {'entier1':2,'texte':'az','entier2':3}
print("nombre d'éléments de c :", len(c))
for i in range(len(c)):
    print(c[i])
```

Mais le dictionnaire étant itérable, nous avons vu que nous pouvions parcourir ses **valeurs**

```python
for val in c.values():
    print(val)
```

Nous aurions aussi pu parcourir ses **clés** (deux méthodes) : 

```python
for cle in c.keys():
    print(cle)
```

```python
for cle in c:
    print(cle)
```

Nous aurions pu aussi parcourir les couples **(cle,valeur)**

```python
for cle,val in c.items():
    print(cle,val)
```

# Complément 3 : **ajout, suppression d'éléments dans un conteneur**

## Ajout
### Dans une liste

```python
# liste
lst = [2, 'az', 5]
print('Avant :', lst)
lst.append(99)             # Ajoute à la fin
print('Après l\'ajout en fin de liste :', lst)
lst.insert(2, 99)         # Insère 99 à l'index 2
print('Après l\'ajout en position 2 :', lst)
```

### Dans un dictionnaire

```python
dic = {'entier1': 2, 'texte': 'az', 'entier2': 3}
print('Avant :', dic)
dic['entier3'] = 'oui'
print('Après l\'ajout :', dic)
```

### Dans un set

```python
ens = {2, 5, 'az'}
print('Avant :', ens)
ens.add(4)
print('Après l\'ajout :',ens)
```

### Dans un tuple
Un tuple n'étant pas muable, l'ajout d'un élément est impossible

## Suppression d'un élément 
### Dans une liste

```python
# liste
lst = [2, 'az', 5,'b',9] 
print('Avant :', lst)
lst.remove('b')    # Suppression de l'élément 'b'
print('Après la suppression de l\'élément \'b\' avec remove :', lst)
lst.pop(2)       # Suppression de l'élément d'indice 2
print('Après la suppression de l\'élément d\'indice 2 avec pop :', lst)
```

**Remarque sur la méthode pop(i)**
* elle modifie la liste, ci-dessus la liste **a**
* mais elle retourne aussi la valeur supprimée de la liste
**Exemple :**

```python
# liste
lst = [2, 'az', 5,'b',9] 
print('Avant :', lst)
val_suppr = lst.pop(2)       # Suppression de l'élément d'indice 2 et enregistrement de cette valeur dans valsuppr
print('Après la suppression de l\'élément d\'indice 2 :', lst)
print('La valeur supprimée de la liste :', val_suppr)
```

### Dans un dictionnaire

```python
dic = {'entier1': 2, 'texte': 'az', 'entier2': 3}
print('Avant :', dic)
del dic['entier1']
print('Après la suppression avec del dic[\'entier1\']:', dic)
val_suppr = dic.pop('texte')
print('Après la suppression avec pop(\'texte\'):', dic)
print('Valeur supprimée :', val_suppr)
```

### Dans un set

```python
ens = {2, 5, 'az'}
print('Avant :', ens)
ens.remove(2)
print('Après la suppression :',ens)
```

**Remarque :** la méthode `remove()` provoque une erreur si l'élément n'est pas dans le set. La méthode `discard()` est équivalente à `remove()`, tout en ne provoquant pas d'erreur si l'élément est absent.
