* ### **Image en noir et blanc**
Le programme suivant permet de créer une image à damiers.

```python
tab4 = np.array([[0 for i in range(n)] for j in range(n)], dtype=np.uint8)
for i in range(n//2) : 
    for j in range(n//2,n) : 
        tab4[i,j] = 255
for i in range(n//2,n) : 
    for j in range(n//2) : 
        tab4[i,j] = 255
plt.imshow(tab4, cmap = 'gray', vmin = 0, vmax = 256)
```

### Remarque : opérateurs de division euclidienne
**n//2** : quotient de la division euclidienne par 2. **n/2** est un float, **n//2** est un entier.  *
**n % 2** : reste de la division euclidienne par 2 (soit 0, soit 1)

## **b. Image en niveaux de gris**

## Remarque préliminaire : nécessité de travailler dans $\mathbb{Z}/256\mathbb{Z}$
Commençons pas executer et commenter le programme suivant : 

```python
n = 1000
t1 = np.array([[j  for j in range(n)] for i in range(n)])
print(t1)
plt.imshow(t1, cmap = 'gray', vmin = 0, vmax = 256)
```

### $\mathbb{Z}/256\mathbb{Z}$
Pour pouvoir être traduit en niveau de gris un entier doit être compris entre 0 et 255.  
Cela signifie aussi que si nous réalisons des opérations sur des pixels, celles-ci doivent se réaliser dans $\mathbb{Z}/256\mathbb{Z}$, ou **modulo 256** ou, en utilisant `%256`, le reste de la division euclidienne par 256,  en Python.

```python
n = 1000
t1 = np.array([[j % 256 for j in range(n)] for i in range(n)])
print(t1)
plt.imshow(t1, cmap = 'gray', vmin = 0, vmax = 256)
```

## **c. Un autre exemple de calcul sur les niveaux de gris**

### Remarque préliminaire


```python
t1 = np.array([[i % 256 for i in range(n)] for j in range(n)])
t2 = np.array([[j % 256 for i in range(n)] for j in range(n)])
# Initialisation de t
t3 = np.array([[0 for i in range(n)] for j in range(n)])
print(type(t3))
for i in range(n) :
    for j in range(n) :
        t3[i,j] = (t1[i,j] + t2[i,j]) % 256
plt.imshow(t3, cmap = 'gray', vmin = 0, vmax = 256)
```
