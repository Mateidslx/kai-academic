## Opérateurs de la division euclidienne 
Sous Python, deux opérateurs permettent d'obtenir les résultats de la **division euclidienne** : 
* **a // b** : donne le quotient de la division euclidienne de a par b 
* **a % b** : donne le reste de la division euclidienne de a par b 

```python
dividende = 14
diviseur = 4
quotient = dividende // diviseur   # // calcule le quotient de la division euclidienne
reste = dividende % diviseur       # % calcule le reste de la division euclidienne
print(f"Dans la division de {dividende} par {diviseur}, {dividende} est le dividende, {diviseur} est le diviseur") 
print(f"Quotient de la division de 14 par 4: {quotient}")
print(f"Reste de la division de 14 par 4: {reste}")
print(f'{dividende} = 4 * {quotient} + {reste}')
```

## **Complexité premier(n)**
La **complexité en temps** d'un algorithme mesure son temps d'exécution en fonction de la taille de l'entrée. 
Cette durée est liée au nombre d'opérations nécessaires pour arriver au résultat.

Le cas de la complexité en **grand O** est le plus grossier. On s'intéresse principalement au nombre d'étapes générées par les boucles et dans le **pire des cas**, ce qui signifie, par exemple, que dans une instruction conditionnelle (**if**), la condition est supposée toujours vraie.   
**Peu importe le nombre d'opérations à chaque étape** : qu'il y ait 5 ou 1000 opérations, la complexité en **grand O** sera la même. *Nous reviendrons sur cette notion en fin d'année (Méthodes numériques).*

Les calculs de complexité peuvent se faire en analysant l'algorithme, ce qui nécessite parfois de mettre en œuvre des outils mathématiques assez avancés. 

Concernant l'algorithme de **premier(n)**, la boucle **for i in range(2, int(n\*\*0.5) + 1) :** comprend $\sqrt n $ (plus précisément la partie entière de $\sqrt n $) étapes et nous avons donc une complexité en $O(\sqrt n)$.

La détermination de la complexité n'est pas toujours aussi simple et on peut aussi analyser l'exécution du programme par la machine pour l'illustrer : 
* Soit en la **chronométrant**.
* Soit en utilisant des **compteurs** dans les boucles.

Nous allons représenter graphiquement les évolutions du temps d'exécution ou du nombre d'itérations en fonction de différentes tailles d'entrée.

### **Préalable : représenter des courbes !**
Ci-dessous, un exemple de tracé permettant de représenter graphiquement la fonction $f(x)=\sqrt x=x^{0,5}$ sur $[0,100]$. 

On utilise la fonction **plot** de la collection **pyplot** de la bibliothèque **matplotlib**.  
L'instruction **plot(x,y)** prend en entrée deux listes **x** et **y** de même longueur et trace, en les reliant, les points de coordonnées **(x[i],y[i])**
### Variables utilisées
* Les listes **x** et **y** doivent avoir une même longueur, que l'on notera **nbPts**
* On représente la courbe pour des valeurs de **x** appartenant à un intervalle dont on précisera la borne inférieure **bInf** et la borne supérieure **bSup**. La longueur de l'intervalle est donc **lenInt = bSup - bInf**
* **delta** est l'écart entre chaque point dans la liste **x**, il se déduit de **nbPts** et **lenInt** : $\frac  {lenInt}{nbPts}$

```python
# Préparation 
bInf = 0                      # Définition des deux bornes de l'intervalle [Binf, Bsup] 
bSup = 1000                   # On représentera la courbe pour des valeurs de x dans cet intervalle
lenInt = bSup - bInf          # longueur de l'intervalle (borne sup moins borne inf)
nbPts = 100                   # On prend nbPts points sur l'intervalle [Binf, Bsup] 
delta = round(lenInt/nbPts,1) # delta représente l'écart entre chaques points dans [Binf, Bsup].

# Préparation des abscisses, par exemple 1000 (nbPts) valeurs entre 0 et 100 : 0, 0.1, 0.2, 0.3,..., 99.9
x = [bInf + delta*i for i in range (nbPts)]
# Préparation des ordonnées : on calcule les images des valeurs précédentes par la fonction racine (x puissance 0,5)
y = [ val_x**0.5 for val_x in x]

print(f"delta : {delta}")
print(f"4 premières valeurs de x : {x[0:4]}")
print(f"4 premières images (racine de x) : {y[0:4]}")
```

#### **Représentation avec plot**
Par défaut, les **nbPts** de la courbe sont reliés entre eux.

```python
import matplotlib.pyplot as plt
plt.plot(x,y)  
plt.show() # affiche la figure à l'écran (inutile en mode interactif, comme le print)
```

### **Chronométrer avec un "timer" - Facultatif (mais intéressant dans certaines situations)**
**timeit** est une des bibliothèques Python qui permet de chronométrer l'exécution de morceaux de programmes.
L'inconvénient de cette mesure du temps réel est que l'exécution peut être perturbée par des éléments indépendants de l'algorithme (processeur, charge...).

### Exemple
Nous testons la durée d'exécution de la fonction **premier(n)** pour diverses valeurs **entières** de **n** que nous représenterons sur l'axe des abscisses.   
* **nMax** : valeur maximale de **n**
* **nbPts** : nombre de valeurs testées
* **delta** : écart entre chaque valeur testée dans $[1,nMax]$
* La liste **x** contient **nbPts** valeurs entières de **n**, qui correspondent aux **abscisses** des points de la courbe
* la liste **yTemps** contient les **nbPts** durées d'exécution de **premier(n)** pour les **nbPts** valeurs de **x**.  
  **yTemps** correspond aux **ordonnées** des points de la courbe

```python
import timeit
nMax = 50000              # Valeur maximale de n
nbPts = 1000              # Nombre de valeurs testées
delta = int(nMax/nbPts)   # Les valeurs de x doivent être entières

# Pour l'axe des x : nbPts valeurs de n, de 50 à 50000, avec un pas de 50 : 50, 100, ..., 49950, 50000
x = [delta * i for i in range(1, nbPts+1)]
yTemps = [0] * nbPts

for i in range(nbPts) : 
    tps1 = timeit.default_timer()  # tps1 : début
    premier(x[i])
    tps2 = timeit.default_timer()  # tps2 : fin
    yTemps[i] = tps2 - tps1       # y_temps[i] : durée d'exécution de premier(x[i])

# Représentation graphique
import matplotlib.pyplot as plt
plt.plot(x,yTemps)
```

### **Introduction d'un compteur dans les boucles**
Dans l'exemple ci-dessous, on introduit un compteur qui s'incrémente à chaque nouvelle itération dans la boucle **for**.   
La liste **yCpt** contient le nombre d'itérations dans la boucle **for** pour chaque valeur **x[i]** de **x**.


```python
nMax = 50000           # Valeur maximale de n
nbPts = 10000             # Nombre de valeurs testées
delta = int(nMax/nbPts)  # Les valeurs de x doivent être entières

# Pour l'axe des x : nbPts valeurs de n, de 0 à 50000, avec un pas de 50 : 50, 100, ..., 49950, 50000
x = [delta * i for i in range(1, nbPts+1)]
yCpt = [0] * nbPts

# Création d'une fonction renvoyant le nombre d'itérations de la boucle de la fonction premier, en fonction de n
def compteurPremier(n):
    prem = True
    compt = 0
    if n<=1 : 
        prem = False
    else : 
        # Parcourir tous les entiers de 2 jusqu'à la racine carrée de n
        for i in range(2, int(n**0.5) + 1) :
            # Nouvelle itération dans la boucle : incrémentation du compteur
            compt += 1
            # Vérifier si i divise n
            if n % i == 0 :
                prem = False
    return compt

# valeurs y_compteur
for i in range(nbPts) : 
    yCpt[i] = compteurPremier(x[i])

# Représentation graphique
plt.plot(x,yCpt)
```

### **Complexité de premier(n) en fonction du nombre de bits utilisés pour stocker n**
Nous nous intéressons maintenant à l'évolution du nombre d'itérations de **premier(n)**, non plus en fonction de **n**, mais en fonction du **nombre de bits** utilisés pour représenter l'entier **n** en binaire.  
En Python, la fonction **n.bit_length()** renvoie ce nombre de bits.

Plus précisément, pour un entier **$n > 0$**, **n.bit_length()** correspond à l'unique entier positif $p$ tel que : $2^{p-1} \le n < 2^p$, ce qui signifie que $ p = \lfloor \log _2 (n) \rfloor + 1$, $\lfloor \log _2 (n) \rfloor$ correspondant à la partie entière de  $\log _2 (n)$.  
Vous avez vu ce résultat dans les ressources informatiques du S1, nous le reverrons en S2 dans la ressource **Méthodes numériques**.  


Dans la nouvelle représentation graphique, nous aurons donc, en **abscisse**, le nombre de **bits** nécessaires pour représenter **n** et en **ordonnées**, le nombre d'itérations dans le calcul de **premier(n)**.
* **nbMax** : valeur maximale de **n**.
* **nbPts** : nombre de valeurs de **n** testées.
* **delta** : écart entre chaque valeur testée dans $[1,nbMax]$.
* La liste **valn** contient **nbPts** valeurs de **n**, séparées d'un intervalle de **delta**.
* La liste **nBits** contient les nombres de bits nécessaires au stockage des entiers de **valn**.
* la liste **yCpt** contient le nombre d'itérations nécessaire au calcul de **premier(n)** pour chaque valeur de **valn**.

On représente une courbe avec, en abscisse, les valeurs de **nBits** et en ordonnée, celles de **yCpt**.

```python
nbMax = 500000
nbPts = 100000           # Nombre de valeurs testées
delta = int(nbMax/nbPts)  # Les valeurs de x doivent être entières

valn = [delta * i for i in range(1, nbPts+1)] # x contient nbPts valeurs entières 
nBits = [n.bit_length() for n in valn]            # b contient les nombres de bits nécessaires au stockage des entiers de x
yCpt = [0] * nbPts

# valeurs y_compteur : on utilise la fonction compteurPremier(n) précédente
# la liste y_compteur est la même que précédemment
for i in range(nbPts) : 
    yCpt[i] = compteurPremier(valn[i])

# Représentation graphique : y_compteur en fonction du nombre de bits (valeurs de b) et non plus de n (valeurs de x)
import matplotlib.pyplot as plt
plt.plot(nBits,yCpt)
```

### Bilan important
L'algorithme de primalité que nous avons utilisé dans la fonction **premier(n)** a une complexité en **O($\sqrt n$)**, mais si on note $p$ le nombre de bits utilisés pour représenter $n$, la complexité est exponentielle, en **O($2^{\frac p 2}$)**.   
Il existe des algorithmes plus efficaces, mais toujours de complexité **suprapolynomiale**, c'est-à-dire plus lente que n'importe quel polynôme. Il en est de même des algorithmes de **décomposition en facteurs premiers**, ce qui a pour conséquence de "protéger" les clés privées dans le chiffrement RSA... Nous en reparlerons en S3.

#### **Tests probabilistes - Facultatif**
Il existe des tests decomplexité **polynomiale** pour la primalité, mais ils sont **probabilistes**, ce qui signifie qu'ils peuvent se **tromper** en désignant comme premiers des entiers qui ne le sont pas.
En Python la fonction **isprime()** de la bibliothèque **sympy** utilise, en partie, un test de primalité **probabiliste** : 
* pour $n < 2^{64}$, le résultat est certain,
* pour $n \ge 2^{64}$, il existe une très faible probabilité qu'un nombre détecté comme premier ne le soit pas.

Nous allons illustrer la notion de **test probabiliste** avec le **test de primalité de Fermat**.   
Le test de primalité de Fermat est basé sur le **petit théorème de Fermat** : 
* il ne se trompe pas en détectant qu'un nombre n'est pas premier (composé)
* en revanche, il peut se tromper en prétendant qu'un nombre est premier.
  
**Rappel du petit théorème de Fermat** : si $n$ est un nombre premier, alors pour tout entier $a$, $n$ divise $a^{n}- a$ ce qui signifie aussi que $a^{n} ≡ a[n]$.   
***Si de plus a est premier avec n et donc inversible, et notamment si $1 \le a \le n-1$, alors $a^{n-1} ≡ 1[n]$***

On appelle **témoin de Fermat** pour $n$, l'entier  $a, 1 \le a \le n-1$ tel que $a^{n-1} \not\equiv 1[n]$.  
L'existence d'un **témoin de Fermat** pour $n$ est la preuve que $n$ **n'est pas premier**.   
Dans le **test de primalité de Fermat**, on teste une ou plusieurs valeurs $a$ choisies au hasard entre $1$ et $n-1$ et on calcule $a^{n-1}[n]$ pour la ou les valeurs de $a$ choisies, **si on obtient toujours $a^{n-1} \equiv 1[n]$, on décide que $n$ est premier**. 
Peut-être à tort, car il peut exister des entiers **n** qui vérifient $a^{n-1} \equiv 1[n]$  et qui ne sont pas premiers, ces entiers sont appelés **pseudo-premiers** (de Fermat) de base a.  

Dans le programme ci-dessous, $k$ représente le nombre de valeurs $a$ choisies au hasard entre $1$ et $n-1$ pour tester $a^{n-1} \equiv 1[n]$.


```python
import random
def testFermat(n, k):
    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    for i in range(k):                  # k : nombre de répétitions du test
        a = random.randint(2, n - 2)    # Choisir un nombre a au hasard entre 2 et n-2
        if pow(a, n - 1, n) != 1:       # a**(n-1)=1 ?
            return False  # Composite
    return True  # Probablement premier
```

```python
testFermat(123456789, 7)
```

## **Algorithme d'Euclide**
**Principe** : si $q$ et $r$ sont respectivement le quotient et le reste de la division euclidienne de $n$ par $m$ ($n \geq m$) alors : $pgcd \left(n,m \right)=pgcd\left( m,r \right)$.  
L'algorithme consiste à effectuer, à chaque étape, la division du diviseur et du reste de l'étape précédente.  
Jusqu'à trouver un reste nul. Le $pgcd$ correspond au dernier reste non nul.

## **Algorithme d'Euclide étendu**	
L'algorithme d'Euclide étendu, permet, non seulement de calculer $p=pgcd(a,b)$, mais aussi $u$ et $v$, les coefficients de Bézout vérifiant $p=a \times u + b \times v$. 
*"étendu" signifiant "étendu au calcul des coefficients de Bézout"*

#### **Quelques rappels sur la méthode itérative**
##### **Principe de l'algorithme**
* Les **étapes** correspondent à celles de l'algorithme d'Euclide : la **division euclidienne** du diviseur par le reste de l'étape précédente. Jusqu'à trouver un reste nul, le $pgcd$ correspondant au dernier reste non nul.
* Avec l'algorithme d'Euclide étendu, on ajoute, à chaque étape, l'expression du reste comme **combinaison linéaire** de $a$ et de $b$.

##### **Notations (mathématiques)**  
$r_n, q_n$ : reste et quotient obtenu à l’étape $n$  
$u_n,v_n$ : coefficients de bézout intermédiaires obtenus à l’étape $n$  
À chaque étape, le reste $r_n$ s'exprime comme combinaison linéaire de $a$ et de $b$, on a donc : 
* $r_n = u_n \times a - v_n \times b$
* Mais l'on a aussi $r_n$ qui est le reste de la division euclidienne des restes des étapes précédentes : $r_n= r_{n-2} -q_n \times r_{n-1}$

${\begin{cases} r_n = u_n \times a - v_n \times b \\ r_{n-1} = u_{n-1} \times a - v_{n-1} \times b\\ r_{n-2} = u_{n-2} \times a - v_{n-2} \times b \end{cases}}$  
$r_n= r_{n-2} -q_n \times r_{n-1}  \Rightarrow  (u_n \times a - v_n \times b)=(u_{n-2} \times a - v_{n-2}) -q_n \times (u_{n-1} \times a - v_{n-1})$  
##### **Conséquence, en factorisant par a et par b, on obtient :**  
$u_n=u_{n-2}-q_n \times u_{n-1}$  
$v_n=u_{n-2}-q_n \times v_{n-1}$

Dans un premier temps, **tab_add_list(n)** renvoie la table sous forme de liste de listes. Le $j^\text {ème}$ élément de la $i^ \text {ème}$ liste correspond à $(i+j)[n]$   
Ci-dessous, on vérifie pour $n=7, i=5$ et $j=6$

```python
# Addition
def tab_add_list(n):
    add=[]
    for i in range(n):
        add.append([])
        for j in range(n):
            add[i].append((i+j)%n)
    return(add)
tab = tab_add_list(7)
print(tab)
print(tab[5][6])      # 5+6[7]
```

### **Amélioration1**
Pour afficher la liste de listes sous forme tabulaire, on peut la transformer en tableau (array) de la bibliothèque **numpy**.

```python
import numpy as np
n=7
tab = tab_add_list(n)
arr = np.array(tab)
print(arr)
```

### **Amélioration2**
Pour afficher la liste de listes sous forme tabulaire, on peut aussi la transformer en dataframe **pandas**.   
Le **dataframe pandas** correspond au tableau **individus x variables** que vous avez déjà vu en statistique (et que nous reverrons très vite en S2 (R2.08 Statistique descriptive).  
Dans un **dataframe**, il est possible, avec les options **index** et **columns** d'ajouter des noms aux lignes et aux colonnes du tableau.

```python
import pandas as pd
n = 7
tab = tab_add_list(n)
df=pd.DataFrame(tab,columns=[i for i in range(n)],index=[i for i in range(n)])    
print(df)
```

```python
def color_cells(val):
    bc = 'background-color: white'
    if val == 0 :
        bc = 'background-color: green'
    return bc

df.style.map(color_cells)
```
